```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
data=pd.read_csv("star_classification.csv")
data
```

```python
data.columns.str.strip()
```

```python
x= data[['u', 'g', 'r', 'i', 'z', 'redshift']]
x
```

```python
y=data[['class']]
y
```

```python
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
model=Sequential()
model.add(Dense(8,input_dim=3,activation="relu"))
model.add(Dense(3,activation='softmax'))
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
model.fit(x,y,epochs=50)
```

```python

Ultraviolet = float(input("Enter Ultraviolet: "))
Green = float(input("Enter Green: "))
Red = float(input("Enter Red: "))
Infrared = float(input("Enter Infrared: "))
NearIR = float(input("Enter NearIR: "))
Redshift = float(input("Enter Redshift: "))
```

```python

test = np.array([[Ultraviolet, Green, Red, Infrared, NearIR, Redshift]])
pred = model.predict(test)[0]
labels=["GALAXY","QSO","STAR"]

print("Prediction:", le.classes_[np.argmax(pred)])
```
